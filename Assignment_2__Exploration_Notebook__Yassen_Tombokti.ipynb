## Downloading the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
from sklearn.linear_model import Ridge, Lasso
import time
from sklearn.model_selection import GridSearchCV
```

## Exploring the Data

```python
df = df_raw.copy()
df.info()
print(df.isna().sum())
print(df[df.isna().any(axis=1)])
print(df.isna().mean().sort_values(ascending=False))
print(len(df), len(df.dropna()))
# since I would lose too much data if I remove nulls, I will fill nulls with the median

df["Courier_Experience_yrs"] = df["Courier_Experience_yrs"].fillna(df["Courier_Experience_yrs"].median())
for col in ["Weather", "Time_of_Day", "Traffic_Level"]:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isna().sum())
```

```python
numerical_cols = [
    "Distance_km",
    "Preparation_Time_min",
    "Courier_Experience_yrs",
    "Delivery_Time_min",
]

print("=== Descriptive Statistics ===")
print(df[numerical_cols].describe().to_string())

print("\n=== Potential Outliers (IQR Method) ===")
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)][col]

    print(f"\n{col}:")
    print(f"  Lower bound: {lower:.2f}")
    print(f"  Upper bound: {upper:.2f}")
    print(f"  Number of potential outliers: {len(outliers)}")
    print(f"  Outlier values: {sorted(outliers.tolist())}")
```

```python
# traffic level remains a single column because each category is more of the same
df["Traffic_Level"] = df["Traffic_Level"].map({"Low": 0, "Medium": 1, "High": 2})
print(df["Traffic_Level"].value_counts(dropna=False))

# the rest of the text data needs to be split into dummy columns because
# the categories are not steps, but actual categories
X = pd.get_dummies(df.drop(columns=["Order_ID", "Delivery_Time_min"]), drop_first=True, dtype=int)
y = df["Delivery_Time_min"]

print(X.shape)             # expect (1000, 13)
print(X.isna().sum().sum())  # expect 0
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# made no difference in accuracy
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Uses ordinary least squares (OLS) by default so I don't to configure that
model = LinearRegression()
model.fit(X_train, y_train)

train_pred = model.predict(X_train)
pred = model.predict(X_test)

train_mae = mean_absolute_error(y_train, train_pred)
test_mae = mean_absolute_error(y_test, pred)

train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
test_rmse = np.sqrt(mean_squared_error(y_test, pred))

print("Training MAE:", train_mae)
print("Test MAE:", test_mae)
print("Training RMSE:", train_rmse)
print("Test RMSE:", test_rmse)
```

```python
plt.scatter(y_test, pred, alpha=0.7)

# Perfect prediction reference line
low = min(y_test.min(), pred.min())
high = max(y_test.max(), pred.max())
plt.plot([low, high], [low, high], "r--", label="Perfect prediction")

plt.xlabel("Real values")
plt.ylabel("Predicted values")
plt.title("Real vs. Predicted Values")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Ridge (L2 penalty), default alpha = 1.0
ridge = Ridge()
ridge.fit(X_train, y_train)
ridge_train_pred = ridge.predict(X_train)
ridge_test_pred = ridge.predict(X_test)

print("Ridge")
print("  Training MAE:", mean_absolute_error(y_train, ridge_train_pred))
print("  Test MAE:", mean_absolute_error(y_test, ridge_test_pred))
print("  Training RMSE:", np.sqrt(mean_squared_error(y_train, ridge_train_pred)))
print("  Test RMSE:", np.sqrt(mean_squared_error(y_test, ridge_test_pred)))

# Lasso (L1 penalty), default alpha = 1.0
lasso = Lasso()
lasso.fit(X_train, y_train)
lasso_train_pred = lasso.predict(X_train)
lasso_test_pred = lasso.predict(X_test)

print("Lasso")
print("  Training MAE:", mean_absolute_error(y_train, lasso_train_pred))
print("  Test MAE:", mean_absolute_error(y_test, lasso_test_pred))
print("  Training RMSE:", np.sqrt(mean_squared_error(y_train, lasso_train_pred)))
print("  Test RMSE:", np.sqrt(mean_squared_error(y_test, lasso_test_pred)))
```

```python
param_grid = {"alpha": np.logspace(-3, 2, 30)}

# 5-fold cross-validation grid search for Ridge
ridge_search = GridSearchCV(Ridge(), param_grid, cv=5,
                            scoring="neg_mean_absolute_error")

start = time.time()
ridge_search.fit(X_train, y_train)
ridge_time = time.time() - start

print("Best alpha:", ridge_search.best_params_["alpha"])
print("Search time (s):", ridge_time)

# Evaluate the best Ridge model
best_ridge = ridge_search.best_estimator_
best_ridge_train_pred = best_ridge.predict(X_train)
best_ridge_test_pred = best_ridge.predict(X_test)

print("Training MAE:", mean_absolute_error(y_train, best_ridge_train_pred))
print("Test MAE:", mean_absolute_error(y_test, best_ridge_test_pred))
print("Training RMSE:", np.sqrt(mean_squared_error(y_train, best_ridge_train_pred)))
print("Test RMSE:", np.sqrt(mean_squared_error(y_test, best_ridge_test_pred)))


lasso_search = GridSearchCV(Lasso(max_iter=10000), param_grid, cv=5,
                            scoring="neg_mean_absolute_error")

start = time.time()
lasso_search.fit(X_train, y_train)
lasso_time = time.time() - start

print("Best alpha:", lasso_search.best_params_["alpha"])
print("Search time (s):", lasso_time)

# Evaluate the best Lasso model
best_lasso = lasso_search.best_estimator_
best_lasso_train_pred = best_lasso.predict(X_train)
best_lasso_test_pred = best_lasso.predict(X_test)

print("Training MAE:", mean_absolute_error(y_train, best_lasso_train_pred))
print("Test MAE:", mean_absolute_error(y_test, best_lasso_test_pred))
print("Training RMSE:", np.sqrt(mean_squared_error(y_train, best_lasso_train_pred)))
print("Test RMSE:", np.sqrt(mean_squared_error(y_test, best_lasso_test_pred)))

coefs = pd.DataFrame({
    "OLS": model.coef_,
    "Ridge": best_ridge.coef_,
    "Lasso": best_lasso.coef_,
}, index=X.columns)
print(coefs.round(2))
```
